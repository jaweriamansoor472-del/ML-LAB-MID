### Introduction

Support Vector Machine (SVM) is a supervised machine learning algorithm used for classification and regression tasks. It works by finding the best boundary that separates different categories of data. The boundary is selected to create the maximum margin between the categories. In this project, SVM is applied to the medical insurance dataset to classify insurance charges into high and low categories based on customer information such as age, BMI, smoking status, and other features.

```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv("insurance.csv")

# Display first five rows
df.head()
```

```python
# Check dataset dimensions
print("Dataset Shape:", df.shape)

# Check column names, data types, and non-null values
print("\nDataset Information:")
df.info()

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Display statistical summary
print("\nStatistical Summary:")
df.describe()
```

df.shape shows the number of rows and columns.



df.info() shows column names and data types.



df.isnull().sum() counts missing values in each column.



df.describe() shows statistics such as mean, minimum, and maximum.

**CREATING THE TARGET VARIABLE**

```python
# Calculate median insurance charges
median_charges = df["charges"].median()

# Classify charges into high and low categories
df["high_charges"] = (df["charges"] > median_charges).astype(int)

# Display results
print("Median Charges:", median_charges)

print("\nTarget Variable Distribution:")
print(df["high_charges"].value_counts())

df.head()
```

Median: The middle value of insurance charges.



0 = Low charges: Charges less than or equal to the median.



1 = High charges: Charges greater than the median.



value_counts() counts how many records belong to each category.
