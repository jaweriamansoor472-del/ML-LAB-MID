```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("insurance.csv")

print(df.head())
```

**HISTOGRAM (AGE DISTRIBUTION)**

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["age"], bins=15, kde=True)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```

What it explains: Shows how the customers in the dataset are distributed across different ages.

**HISTOGRAM (BMI DISTRIBUTION)**

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["bmi"], bins=20, kde=True)

plt.title("BMI Distribution")
plt.xlabel("BMI")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```

What it explains: Shows the distribution of customers' BMI values.

**HISTOGRAM (INSURANCE CHARGES DISTRIBUTION)**

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["charges"], bins=30, kde=True)

plt.title("Distribution of Insurance Charges")
plt.xlabel("Insurance Charges")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

What it explains: Shows how insurance costs are distributed and whether there are unusually high-cost cases.

**BAR CHART (SMOKER vs. AVERAGE CHARGES)**

```python
average_charges = df.groupby("smoker")["charges"].mean()

plt.figure(figsize=(7, 5))

average_charges.plot(kind="bar")

plt.title("Average Insurance Charges by Smoking Status")
plt.xlabel("Smoker")
plt.ylabel("Average Insurance Charges")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

What it explains: Compares the average insurance cost of smokers and non-smokers.

**REGION vs. AVERAGE CHARGES**

**CHILDREN vs. AVERAGE CHARGES**

```python
average_charges = df.groupby("children")["charges"].mean()

plt.figure(figsize=(8, 5))

average_charges.plot(kind="bar")

plt.title("Average Insurance Charges by Number of Children")
plt.xlabel("Number of Children")
plt.ylabel("Average Insurance Charges")
plt.tight_layout()
plt.show()
```

**AGE GROUP vs. AVERAGE CHARGES**

```python
df["age_group"] = pd.cut(
    df["age"],
    bins=[17, 25, 35, 45, 55, 65, 100],
    labels=["18-25", "26-35", "36-45", "46-55", "56-65", "66+"]
)

average_charges = df.groupby(
    "age_group",
    observed=False
)["charges"].mean()

plt.figure(figsize=(9, 5))

average_charges.plot(kind="bar")

plt.title("Average Insurance Charges by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Average Insurance Charges")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

What it explains: Makes it easier to compare insurance costs across different age groups.

**CORRELATION HEATMAP**

```python
numeric_columns = [
    "age",
    "bmi",
    "children",
    "charges"
]

correlation = df[numeric_columns].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

What it explains: Shows the strength and direction of relationships between the numerical variables.

**PAIR PLOT (NUMERICAL VARIABLES)**

```python
sns.pairplot(
    df[["age", "bmi", "children", "charges"]]
)

plt.suptitle(
    "Pairwise Relationships of Numerical Variables",
    y=1.02
)

plt.show()
```

What it explains: Shows distributions and relationships between age, BMI, children, and charges simultaneously.
